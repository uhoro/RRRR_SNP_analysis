## Clonality analysis

### Prepare the variant data

Before this step, make sure to convert vcf variant data to csv file using codes in "clonality.rmd"

```python
import pandas as pd
import os

out_dir = "outputs/"
data_name = out_dir + "/bindedp4_set240924_filt250821_INDEL"

sample_vec = ["SM44B", "C1C5v2", "C1A10", "C1E4", "C2F9", 
                "SM60B", "C3D7v2", "C1C51", "C2H11",
                "C1G11", "C1A1"]
data_csv = data_name + ".csv"
data = pd.read_csv(data_csv)
data = data.drop(columns=['C1A4', 'DPC1A4'])
data_trim = data.replace({float('nan'): pd.NaT})
data_trim = data_trim.dropna(subset=sample_vec)

    
for colname in list(data_trim):
    data_trim[colname] = data_trim[colname].astype(str)
data_trim = data_trim.reset_index(drop=True)

out_fname = out_dir + "filt250821_INDEL"
data_trim_fname = out_fname + "_trim_input.csv"
data_trim.to_csv(data_trim_fname, index = False)
```

### Compute clonality 

Clonality can be computed for each pair of samples using the following python scripts, but it takes time for SNP data especially. 

> python get_pair_overlap_250827.py --data_fname INPUT_CSV_PATH --s1 SAMPLE1_TO_PROCESS --s2 SAMPLE2_TO_PROCESS --pair_overlap_dir OUTPUT_DIR_PATH

It can also be run in the computer cluster using "test_arr_get_pair_overlap.sh"

Make sure to update the working directory etc in test_arr_get_pair_overlap.sh before runnig.

To obtain summary table given the clonality data of each sample pair,

> python get_clonality_summary_250827.py --data_fname data/filt250821_trim_input.csv --pair_overlap_dir result/filt250821_trim --summary_dir_fname result/summary/filt250821_trim_summary
